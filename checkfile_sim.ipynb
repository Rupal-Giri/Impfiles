```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
c = pd.read_csv('checkfile.txt',sep= '\s+',skiprows=0,header= None)
c.columns = ['file','flag','comment']
c
```

```python
c.flag = '+1cx'
c.flag
```

```python
flag = c.flag
len(flag)
```

```python
import numpy as np
import os  
os.makedirs("flags", exist_ok=True)

for n in range(len(flag)):
    i = np.random.randint(0, len(flag), size=n)
    flag_i = flag.copy()
    
    for j in i:
        flag_i[j] = "-1cx"  

    filename = f"flag_{n}.txt"
    filepath = f"flags/{filename}"  

    np.savetxt(filepath, flag_i, fmt="%s") 
    print(f"Saved {filepath}: {flag_i}")


    
```
