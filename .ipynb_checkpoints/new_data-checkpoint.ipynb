```python
def average_every_n_points(df, n):
    df = df.reset_index(drop=True)  # reset index for clean grouping
    # Create a group label for every n rows
    groups = (df.index // n)
    # Group by these labels and take mean of numeric columns
    df_avg = df.groupby(groups).mean()
    return df_avg.reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.lines import Line2D

# === Sources and B dataset name mapping ===
sources = ["1ES2344+514", "1ES1959+650", "3C273", "0954+658", "1334-127", "OJ287"]

b_name_map = {
    "3C273": "1226+023"
}

# === Frequency bands and colors ===
freq_bands = {
    '5GHz': (4000, 7000),
    '8GHz': (7000, 9000),
    '20GHz': (15000, 30000),
    '43GHz': (30000, 50000)
}
colors = {
    '5GHz': 'orange',
    '8GHz': 'forestgreen',
    '20GHz': 'dodgerblue',
    '43GHz': 'navy'
}

# === Load data for all sources ===
source_data_dict = {}
source_data_new_dict = {}
source_data_b_dict = {}

for src in sources:
    source_data_dict[src] = get_source_data(data, src)
    source_data_new_dict[src] = get_source_data(data_NEW, src)
    b_src = b_name_map.get(src, src)
    source_data_b_dict[b_src] = get_source_data(bach_data, b_src)

    # Ensure numeric columns
    for df in [source_data_dict[src], source_data_new_dict[src], source_data_b_dict[b_src]]:
        for col in ['Freq', '#MJD', 'Flux', 'e_Flux']:
            if col in df.columns:
                df.loc[:, col] = pd.to_numeric(df[col], errors='coerce')
        if 'Flux err' in df.columns:
            df.loc[:, 'Flux err'] = pd.to_numeric(df['Flux err'], errors='coerce')
        if 'MJD' in df.columns:
            df.loc[:, 'MJD'] = pd.to_numeric(df['MJD'], errors='coerce')

# === Create subplots ===
fig, axes = plt.subplots(3, 2, figsize=(14, 14), sharex=True)
axes = axes.flatten()

for i, src in enumerate(sources):
    ax = axes[i]
    ax.set_title(src, fontsize=11)

    b_src = b_name_map.get(src, src)

    # Dynamic y-axis limits
    all_fluxes = pd.concat([
        source_data_dict[src]['Flux'] if not source_data_dict[src].empty else pd.Series(),
        source_data_new_dict[src]['Flux'] if not source_data_new_dict[src].empty else pd.Series(),
        source_data_b_dict[b_src]['Flux'] if not source_data_b_dict[b_src].empty else pd.Series()
    ])
    if not all_fluxes.empty:
        ax.set_ylim(0, all_fluxes.max() * 1.2)

    for band, (low, high) in freq_bands.items():
        # Old data
        f_data = source_data_dict[src][source_data_dict[src]['Freq'].between(low, high)]
        if not f_data.empty:
            ax.errorbar(f_data['#MJD'], f_data['Flux'], yerr=f_data['e_Flux'],
                        fmt='o', color=colors[band], alpha=0.6, markersize=3)
            ax.plot(f_data['#MJD'], f_data['Flux'], color=colors[band],
                    alpha=0.5, linestyle=':', linewidth=1)

        # New data
        f_data_new = source_data_new_dict[src][source_data_new_dict[src]['Freq'].between(low, high)]
        if not f_data_new.empty:
            ax.errorbar(f_data_new['#MJD'], f_data_new['Flux'], yerr=f_data_new['e_Flux'],
                        fmt='o', color=colors[band], alpha=0.8, markersize=3)
            ax.plot(f_data_new['#MJD'], f_data_new['Flux'], color=colors[band],
                    alpha=0.5, linestyle=':', linewidth=1)
            

        # B data
        f_data_b = source_data_b_dict[b_src][source_data_b_dict[b_src]['Freq'].between(low, high)]
        if not f_data_b.empty:
            ax.errorbar(f_data_b['MJD'], f_data_b['Flux'], yerr=f_data_b['Flux err'],
                        fmt='o', color=colors[band], alpha=0.8, markersize=3)
           
    ax.grid(True, linestyle=':', alpha=0.5)
    if i >= 4:
        ax.set_xlabel("MJD")
    if i % 2 == 0:
        ax.set_ylabel("Flux density (Jy)")

# === Custom legend for markers (datasets) and colors (frequency) ===


freq_handles = [Line2D([0], [0], color=c, lw=3, label=band) for band, c in colors.items()]

# Place legends outside plot
#fig.legend(handles=dataset_handles, title="Dataset", loc="upper left")
fig.legend(handles=freq_handles, title="Frequency", loc="upper right")

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.suptitle("Flux Density vs. MJD for Six Sources", fontsize=14, y=1.02)
plt.savefig("images/new_data/flux_vs_mjd_6sources.png", dpi=300)
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller, kpss

def stationarity_test(series):
    """
    Perform stationarity tests on the given time series data.
    Returns a dictionary with ADF and KPSS p-values,
    or None if the series is too short.
    """
    cleaned = series.dropna()

    if len(cleaned) < 20:
        return {"adf": None, "kpss": None, "reason": "Too few data points"}

    try:
        adf_result = adfuller(cleaned)
    except ValueError as e:
        return {"adf": None, "kpss": None, "reason": f"ADF error: {e}"}

    try:
        kpss_result = kpss(cleaned, regression='ct')
    except ValueError as e:
        return {"adf": adf_result[1], "kpss": None, "reason": f"KPSS error: {e}"}

    return {"adf": adf_result[1], "kpss": kpss_result[1]}
```

```python


# Lists to store source names with p-values
sources_with_p_values_5 = []
sources_with_p_values_8 = []
sources_with_p_values_20 = []

for i in source_names:
    dat = get_source_data(data_with_aliases_replaced, i)
    dat['Freq'] = pd.to_numeric(dat['Freq'], errors='coerce')
    dat_5 = dat[(dat['Freq'] >= 4000) & (dat['Freq'] <= 6000)]
    dat_8 = dat[(dat['Freq'] >= 7000) & (dat['Freq'] <= 9000)]
    dat_20 = dat[(dat['Freq'] >= 15000) & (dat['Freq'] <= 30000)]
    f_5 = dat_5['Flux']
    f_8 = dat_8['Flux']
    f_20 = dat_20['Flux']
    p_5= stationarity_test(f_5)
    p_8= stationarity_test(f_8)
    p_20= stationarity_test(f_20)

    # Append source names with their p-values
    sources_with_p_values_5.append(f"{i}: {p_5}")
    sources_with_p_values_8.append(f"{i}: {p_8}")
    sources_with_p_values_20.append(f"{i}: {p_20}")

```

```python
import numpy as np
import matplotlib.pyplot as plt

ra_deg = basic['RA'].to_numpy()
dec_deg = basic['Dec'].to_numpy()

# Rotation matrix (ICRS/J2000 -> Galactic)
R = np.array([
    [-0.0548755604, -0.8734370902, -0.4838350155],
    [ 0.4941094279, -0.4448296300,  0.7469822445],
    [-0.8676661490, -0.1980763734,  0.4559837762]
])

def radec_to_galactic(ra_deg, dec_deg):
    ra = np.radians(np.asarray(ra_deg, dtype=float))
    dec = np.radians(np.asarray(dec_deg, dtype=float))
    x_eq = np.cos(dec) * np.cos(ra)
    y_eq = np.cos(dec) * np.sin(ra)
    z_eq = np.sin(dec)
    vec_eq = np.vstack([x_eq, y_eq, z_eq])
    vec_gal = R.dot(vec_eq)
    xg, yg, zg = vec_gal
    l_rad = np.arctan2(yg, xg)
    b_rad = np.arcsin(zg / np.linalg.norm(vec_gal, axis=0))
    l_deg = np.degrees(l_rad)
    l_deg = ((l_deg + 180) % 360) - 180
    b_deg = np.degrees(b_rad)
    return l_deg, b_deg

def plot_in_galactic_no_astropy(ra_deg, dec_deg, labels=None,
                                projection='mollweide',
                                title="Sky plot (Galactic coordinates)"):
    ra = np.asarray(ra_deg, dtype=float)
    dec = np.asarray(dec_deg, dtype=float)
    if ra.shape != dec.shape:
        raise ValueError("ra_deg and dec_deg must have the same shape")
    l_deg, b_deg = radec_to_galactic(ra, dec)
    l_rad = np.radians(l_deg)
    b_rad = np.radians(b_deg)

    fig = plt.figure(figsize=(9,5))
    ax = fig.add_subplot(111, projection=projection)
    ax.set_title(title)
    ax.scatter(l_rad, b_rad, s=30)
    if labels is not None:
        for xi, yi, lab in zip(l_rad, b_rad, labels):
            ax.annotate(str(lab), (xi, yi), fontsize=8, xytext=(3,3), textcoords='offset points')
    ax.grid(True)
    tick_deg = np.arange(-150, 181, 30)
    tick_rad = np.radians(tick_deg)
    tick_labels = [f"{int(t)}\u00b0" for t in tick_deg]
    ax.set_xticks(tick_rad)
    ax.set_xticklabels(tick_labels)
    yticks_deg = np.arange(-60, 61, 30)
    ax.set_yticks(np.radians(yticks_deg))
    ax.set_yticklabels([f"{int(t)}\u00b0" for t in yticks_deg])
    ax.invert_xaxis()
    plt.tight_layout()
    plt.show()
    return fig

```

```python
import pandas as pd
import numpy as np

def hms_to_deg(hms):
    """Convert 'hh:mm:ss' or datetime.time to degrees."""
    if pd.isna(hms):
        return np.nan
    if isinstance(hms, str):
        h, m, s = map(float, hms.split(':'))
    elif hasattr(hms, 'hour'):
        h, m, s = hms.hour, hms.minute, hms.second
    else:
        return float(hms)
    return (h + m/60 + s/3600) * 15  # 24h = 360° → 1h = 15°

def dms_to_deg(dms):
    """Convert '±dd:mm:ss' strings or datetime.time objects to decimal degrees."""
    if pd.isna(dms):
        return np.nan
    if isinstance(dms, str):
        # Normalize Unicode minus characters
        dms = dms.replace('−', '-').replace('–', '-').replace('—', '-').strip()
        parts = dms.split(':')
        sign = -1 if parts[0].startswith('-') else 1
        d = abs(float(parts[0]))
        m = float(parts[1]) if len(parts) > 1 else 0
        s = float(parts[2]) if len(parts) > 2 else 0
    elif hasattr(dms, 'hour'):
        d, m, s, sign = dms.hour, dms.minute, dms.second, 1
    else:
        return float(dms)
    return sign * (d + m/60 + s/3600)

    # Replace 'RA' and 'Dec' with your actual column names
basic['RA_deg'] = basic['RA'].apply(hms_to_deg)
basic['Dec_deg'] = basic['Dec'].apply(dms_to_deg)

basic['RA_deg']
```

```python
def plot_colored_galactic(basic, ra_col='RA_deg', dec_col='Dec_deg', class_col='Blazar type',
                          projection='mollweide', title='Sky Map by Source Type',
                          colors=None, markers=None, save_path=None, dpi=300):
    """
    Plot sources in Galactic coordinates, colored and shaped by classification.
    Automatically removes rows with missing (NaN) classifications.
    
    colors: dict mapping class -> color
    markers: dict mapping class -> marker
    save_path: str or None, if provided, saves the figure to this file
    dpi: resolution for saved figure
    """
    import matplotlib.pyplot as plt
    import numpy as np
    
    df = basic.dropna(subset=[class_col]).copy()
    
    # Convert RA/Dec to galactic l,b
    l_deg, b_deg = radec_to_galactic(df[ra_col], df[dec_col])
    df['l_deg'] = l_deg
    df['b_deg'] = b_deg
    
    classes = df[class_col].unique()
    
    fig = plt.figure(figsize=(9,5))
    ax = fig.add_subplot(111, projection=projection)
    ax.set_title(title)
    
    default_colors = plt.cm.get_cmap('Set1', len(classes)).colors
    default_markers = ['o', 's', '^', 'D', 'v', '<', '>', 'p', '*', 'h', 'H', '+', 'x']
    
    if colors is None:
        colors = {cls: default_colors[i % len(default_colors)] for i, cls in enumerate(classes)}
    if markers is None:
        markers = {cls: default_markers[i % len(default_markers)] for i, cls in enumerate(classes)}
    
    for cls in classes:
        mask = df[class_col] == cls
        ax.scatter(np.radians(df.loc[mask, 'l_deg']),
                   np.radians(df.loc[mask, 'b_deg']),
                   label=str(cls),
                   s=222,
                   alpha=0.5,
                   color=colors[cls],
                   marker=markers[cls],
                   edgecolor='k')
    
    ax.grid(True)
    tick_deg = np.arange(-150, 181, 30)
    tick_rad = np.radians(tick_deg)
    tick_labels = [f"{int(t)}°" for t in tick_deg]
    ax.set_xticks(tick_rad)
    ax.set_xticklabels(tick_labels)
    yticks_deg = np.arange(-60, 61, 30)
    ax.set_yticks(np.radians(yticks_deg))
    ax.set_yticklabels([f"{int(t)}°" for t in yticks_deg])
    ax.invert_xaxis()
    
    ax.legend(title=class_col, loc='lower left', fontsize=8)
    plt.tight_layout()
    
    # Save the figure if save_path is provided
    if save_path is not None:
        plt.savefig(save_path, dpi=dpi, bbox_inches='tight')
        print(f"Figure saved to: {save_path}")
    
    plt.show()

plot_colored_galactic(
    basic,
    class_col='Blazar type',
    colors=custom_colors,
    markers=custom_markers,
    title='Sky Distribution of Blazars',
    save_path='images/Poster/skymap.png',
    dpi=300
)
```
